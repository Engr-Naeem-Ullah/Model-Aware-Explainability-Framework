```python
import pandas as pd
import os

# ===============================
# 1. Load dataset
# ===============================
data_path = r"E:\Abroad period research\New idea for 2026\yield_2006-24.csv"
df = pd.read_csv(data_path)

# ===============================
# 2. Sort chronologically by year
# ===============================
df = df.sort_values(by="year").reset_index(drop=True)

# ===============================
# 3. Determine chronological split
# ===============================
unique_years = sorted(df["year"].unique())

split_ratio = 0.8
split_index = int(len(unique_years) * split_ratio)

train_years = unique_years[:split_index]
test_years = unique_years[split_index:]

# ===============================
# 4. Create train and test sets
# ===============================
train_df = df[df["year"].isin(train_years)]
test_df = df[df["year"].isin(test_years)]

# ===============================
# 5. Save files in same folder
# ===============================
base_dir = os.path.dirname(data_path)

train_path = os.path.join(base_dir, "yield_train.csv")
test_path = os.path.join(base_dir, "yield_test.csv")

train_df.to_csv(train_path, index=False)
test_df.to_csv(test_path, index=False)

# ===============================
# 6. Print summary
# ===============================
print("Chronological split completed successfully.")
print(f"Training years: {min(train_years)} - {max(train_years)}")
print(f"Testing years: {min(test_years)} - {max(test_years)}")
print(f"Training samples: {train_df.shape[0]}")
print(f"Testing samples: {test_df.shape[0]}")
print(f"Saved files:")
print(train_path)
print(test_path)
```
